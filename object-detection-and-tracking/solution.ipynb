# Listing 5.1: Feature-Based Object Representation Using DAISY Descriptors

```python
import cv2
import numpy as np
from skimage.feature import daisy
import matplotlib.pyplot as plt
```

```python
image_path='sample.png'
```

```python
# Load the image in grayscale
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Compute DAISY feature descriptors
descs, descs_img = daisy(image,
    step = 180,
    radius = 58,
    rings = 2,
    histograms = 6,
    orientations = 8,
    visualize=True)

# Convert the descriptors image from grayscale to RGB
# descs_img = cv2.cvtColor(descs_img,cv2.COLOR_GRAY2RGB)
```

```python
# Create subplots for input and output visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Display the original input image
ax1.imshow(image, cmap='gray')
ax1.set_title("Input Image")
ax1.axis("off")

# Display the DAISY descriptors visualization
ax2.imshow(descs_img)
ax2.set_title(f"{descs.shape[0] * descs.shape[1]} DAISY descriptors extracted")
ax2.axis("off")

plt.tight_layout()
plt.show()
```

# Listing 5.2: Template Matching Using OpenCV

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
# Load the image and the template
image = cv2.imread('cocacola.png',0)
template = cv2.imread('cocacola-logo.png',0)
w, h = template.shape[: : -1]

# Perform template matching
res = cv2.matchTemplate(image,
    template,
    cv2.TM_CCOEFF_NORMED)

# Set a threshold and find where
# the matching result exceeds it
threshold = 0.215
loc = np.where(res >= threshold)

# Create a copy of the original image
# in color to draw rectangles
color_image = cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)

# Draw rectangles around the matched regions
for pt in zip(* loc[::-1]):
    cv2.rectangle(color_image,
        pt,
        (pt[0] + w, pt[1] + h),
        (25, 25, 25),
        1)

plt.imshow(color_image)
plt.title('Detected ananas')
plt.show()
```
